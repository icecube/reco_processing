## Configuration
Set `model` and `selected_keys` before running the cells below.

```python
# ---- user parameters ----
BASE_GRAPHS_PATH = os.path.join(THIS_DIR, "graphs")
BASE_SAVE_PATH   = os.path.join(THIS_DIR, "plots")  # per-scan subfolders created automatically

# Select which plots to produce.
# Plots are saved to BASE_SAVE_PATH/{scan_name}/{key}.png
selected_keys = GROUP_ALL
# selected_keys = GROUP_SEPARATE_PLOTTING        # fit parametrization — Astro+Conv+Prompt+Muon components
# selected_keys = GROUP_SEPARATE_NOSYST          # same, best fit without the Snowstorm gradient shift (_nosyst)
# selected_keys = GROUP_SEPARATE_FLAVOR          # 3-flavor parametrization (_3flavor graph) — NuE+NuMu+NuTau+Conv+Prompt+Muon
# selected_keys = GROUP_SEPARATE_FLAVOR_NOSYST   # same, best fit without the Snowstorm gradient shift (_nosyst)
# selected_keys = GROUP_OLD_PASSING_FRACTIONS         # fit parametrization, old passing fractions (data_v3, old gradients)
# selected_keys = GROUP_OLD_PASSING_FRACTIONS_NOSYST  # same, best fit without the Snowstorm gradient shift (_nosyst)
# selected_keys = GROUP_ICE_CRYSTAL              # fit parametrization, ice_crystal fixed to 0.9 / 1.0 / 1.1 / 1.2
# selected_keys = GROUP_ICE_CRYSTAL_NOSYST       # same, without the Snowstorm gradient shift (identical to GROUP_SEPARATE_NOSYST)

# Which plot variants to make (all variants stay defined in plot_configs.py).
# Add False to also make the plots without counts in the legend / without data points.
SHOW_COUNTS = [True]
PLOT_DATA   = [True]

# Draw the conv/prompt parts of the combined muon template (tracks only).
# With True the plots are saved as {key}_muonsplit.png
PLOT_MUON_SPLIT = True

# Additionally make every selected tracks plot without the muon conv/prompt split (saved as {key}.png),
# both with and without the Snowstorm gradient shift (_nosyst)
TRACKS_WITHOUT_MUON_SPLIT = True
```

## Load histograms

```python
step0_path = BASE_GRAPHS_PATH          # flat structure (all models)
save_path  = BASE_SAVE_PATH
os.makedirs(save_path, exist_ok=True)

active_plots = [
    {**p, "plot_muon_split": PLOT_MUON_SPLIT} for p in PLOTS
    if p["key"] in selected_keys and p["show_counts"] in SHOW_COUNTS and p["plot_data"] in PLOT_DATA
]
if TRACKS_WITHOUT_MUON_SPLIT:
    plots_by_key = {p["key"]: p for p in PLOTS}
    track_keys = sorted({
        base + syst
        for p in active_plots if "_channel-tracks_" in p["key"]
        for base in [p["key"].removesuffix("_nosyst")]
        for syst in ["", "_nosyst"]
    })
    active_plots += [{**plots_by_key[k], "plot_muon_split": False} for k in track_keys]
print(f"{len(active_plots)} plots selected")

histogram_collection = load_histograms(step0_path, active_plots)
```

## Make plots

```python
make_all_plots(active_plots, histogram_collection, save_path=save_path, show=False)
```
